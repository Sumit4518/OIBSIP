````python
# ============================================
# STEP 7.1: FEATURES AND TARGET
# ============================================

X = df[numeric_columns]
y = df["species"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
````

````python
# ============================================
# STEP 7.2: TRAIN-TEST SPLIT
# ============================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
````

````python
# ============================================
# STEP 7.3: VERIFY CLASS DISTRIBUTION
# ============================================

print("Training distribution:")
print(y_train.value_counts())

print("\nTesting distribution:")
print(y_test.value_counts())
````

````python
# ============================================
# STEP 7.4: FEATURE SCALING
# ============================================

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Training data scaled successfully.")
print("Testing data scaled successfully.")
````

````python
# ============================================
# STEP 7.5: VERIFY SCALING
# ============================================

print("Training feature means:")
print(X_train_scaled.mean(axis=0))

print("\nTraining feature standard deviations:")
print(X_train_scaled.std(axis=0))
````

````python
# ============================================
# STEP 7.6: PREPROCESSING PIPELINE
# ============================================

from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=1000))
])

print("Pipeline created successfully.")
````

## ⚙️ Data Preprocessing Summary

The dataset was divided into training and testing subsets using an
80:20 split. Stratified sampling was applied to preserve the class
distribution of the three Iris species.

Feature scaling was performed using StandardScaler. The scaler was
fitted only on the training data and then applied to the test data
to prevent data leakage.

A Scikit-learn Pipeline was also created to combine preprocessing
and model training into a reproducible workflow.

````python
# ============================================
# STEP 8.1: IMPORT ML MODELS
# ============================================

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

print("ML models imported successfully!")
````

````python
# ============================================
# STEP 8.2: CREATE ML MODELS
# ============================================

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),

    "K-Nearest Neighbors": KNeighborsClassifier(n_neighbors=5),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42
    ),

    "Support Vector Machine": SVC(
        kernel="rbf",
        probability=True,
        random_state=42
    )
}

print("Models created:")
for name in models:
    print("-", name)
````

````python
# ============================================
# STEP 8.3: TRAIN MODELS
# ============================================

trained_models = {}

for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    trained_models[name] = model

print("All models trained successfully!")
````

````python
# ============================================
# STEP 8.4: GENERATE PREDICTIONS
# ============================================

predictions = {}

for name, model in trained_models.items():
    predictions[name] = model.predict(X_test_scaled)

print("Predictions generated for all models!")
````

````python
# Compare actual vs predicted values

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": predictions["Random Forest"]
})

comparison.head(10)
````

## 🤖 Machine Learning Models

Five classification algorithms were trained and evaluated:

1. Logistic Regression
2. K-Nearest Neighbors
3. Decision Tree
4. Random Forest
5. Support Vector Machine

Using multiple algorithms allows us to compare different modeling
approaches rather than relying on a single classifier.

The models were trained using the training dataset and evaluated
later on the unseen test dataset.

````python
# ============================================
# STEP 9.1: IMPORT EVALUATION METRICS
# ============================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

print("Evaluation metrics imported successfully!")
````

````python
# ============================================
# STEP 9.2: MODEL EVALUATION
# ============================================

results = []

for name in trained_models:

    y_pred = predictions[name]

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(
        y_test,
        y_pred,
        average="weighted"
    )
    recall = recall_score(
        y_test,
        y_pred,
        average="weighted"
    )
    f1 = f1_score(
        y_test,
        y_pred,
        average="weighted"
    )

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    })

results_df = pd.DataFrame(results)

results_df
````

````python
# ============================================
# STEP 9.3: FORMAT RESULTS
# ============================================

results_df[
    ["Accuracy", "Precision", "Recall", "F1-Score"]
] = results_df[
    ["Accuracy", "Precision", "Recall", "F1-Score"]
].round(4)

results_df
````

````python
# ============================================
# STEP 9.4: RANK MODELS
# ============================================

ranked_models = results_df.sort_values(
    by="F1-Score",
    ascending=False
).reset_index(drop=True)

ranked_models
````

````python
# ============================================
# STEP 9.5: BEST MODEL
# ============================================

best_model_name = ranked_models.loc[0, "Model"]
best_f1_score = ranked_models.loc[0, "F1-Score"]

print("Best Model:", best_model_name)
print("Best F1-Score:", best_f1_score)
````

````python
# ============================================
# STEP 9.6: MODEL COMPARISON
# ============================================

metrics_to_plot = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1-Score"
]

plot_df = results_df.set_index("Model")[metrics_to_plot]

ax = plot_df.plot(
    kind="bar",
    figsize=(14, 7)
)

plt.title("Machine Learning Model Performance Comparison")
plt.xlabel("Model")
plt.ylabel("Score")
plt.ylim(0, 1.1)
plt.xticks(rotation=30, ha="right")
plt.legend(title="Metrics")
plt.tight_layout()
plt.show()
````

````python
# ============================================
# STEP 9.8: CONFUSION MATRICES
# ============================================

fig, axes = plt.subplots(
    2, 3,
    figsize=(16, 10)
)

axes = axes.flatten()

for ax, (name, y_pred) in zip(
    axes,
    predictions.items()
):

    cm = confusion_matrix(
        y_test,
        y_pred,
        labels=iris.target_names
    )

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=iris.target_names,
        yticklabels=iris.target_names,
        ax=ax
    )

    ax.set_title(name)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("Actual")

# Hide unused subplot
axes[-1].axis("off")

plt.suptitle(
    "Confusion Matrix Comparison",
    fontsize=16
)

plt.tight_layout()
plt.show()
````

````python
# ============================================
# STEP 10.1: CROSS-VALIDATION
# ============================================

from sklearn.model_selection import StratifiedKFold, cross_val_score

print("Cross-validation tools imported successfully!")
````

````python
# ============================================
# STEP 10.2: STRATIFIED K-FOLD
# ============================================

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

print("5-Fold Stratified Cross-Validation created.")
````

````python
# ============================================
# STEP 10.3: MODEL PIPELINES
# ============================================

from sklearn.pipeline import Pipeline

cv_models = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=1000))
    ]),

    "K-Nearest Neighbors": Pipeline([
        ("scaler", StandardScaler()),
        ("model", KNeighborsClassifier(n_neighbors=5))
    ]),

    "Decision Tree": Pipeline([
        ("scaler", StandardScaler()),
        ("model", DecisionTreeClassifier(random_state=42))
    ]),

    "Random Forest": Pipeline([
        ("scaler", StandardScaler()),
        ("model", RandomForestClassifier(
            n_estimators=100,
            random_state=42
        ))
    ]),

    "Support Vector Machine": Pipeline([
        ("scaler", StandardScaler()),
        ("model", SVC(
            kernel="rbf",
            probability=True,
            random_state=42
        ))
    ])
}
````

````python
# ============================================
# STEP 10.4: CROSS-VALIDATION SCORES
# ============================================

cv_results = []

for name, model in cv_models.items():

    scores = cross_val_score(
        model,
        X,
        y,
        cv=cv,
        scoring="accuracy"
    )

    cv_results.append({
        "Model": name,
        "Mean CV Accuracy": scores.mean(),
        "Std CV Accuracy": scores.std()
    })

cv_results_df = pd.DataFrame(cv_results)

cv_results_df[
    ["Mean CV Accuracy", "Std CV Accuracy"]
] = cv_results_df[
    ["Mean CV Accuracy", "Std CV Accuracy"]
].round(4)

cv_results_df
````

````python
# ============================================
# STEP 10.6: RANK CROSS-VALIDATION RESULTS
# ============================================

cv_ranked = cv_results_df.sort_values(
    by="Mean CV Accuracy",
    ascending=False
).reset_index(drop=True)

cv_ranked
````

````python
# ============================================
# STEP 10.7: CROSS-VALIDATION VISUALIZATION
# ============================================

plt.figure(figsize=(12, 6))

sns.barplot(
    data=cv_ranked,
    x="Model",
    y="Mean CV Accuracy"
)

plt.title("5-Fold Cross-Validation Accuracy")
plt.xlabel("Model")
plt.ylabel("Mean Accuracy")
plt.ylim(0, 1.05)
plt.xticks(rotation=30, ha="right")

plt.tight_layout()
plt.show()
````

## 🔄 Cross-Validation Analysis

To reduce dependence on a single train-test split, Stratified
5-Fold Cross-Validation was performed on all candidate models.

Stratification was used to maintain the distribution of the three
Iris species across the folds.

The mean cross-validation accuracy represents the average model
performance across the five validation folds, while the standard
deviation indicates the consistency of the model.

Cross-validation results will be considered together with the
independent test-set results when selecting the final model.

````python
# ============================================
# STEP 11.1: HYPERPARAMETER TUNING
# ============================================

from sklearn.model_selection import GridSearchCV

print("GridSearchCV imported successfully!")
````

````python
# ============================================
# STEP 11.2: LOGISTIC REGRESSION TUNING
# ============================================

lr_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=2000))
])

lr_params = {
    "model__C": [0.01, 0.1, 1, 10, 100],
    "model__solver": ["lbfgs", "liblinear"]
}

lr_grid = GridSearchCV(
    estimator=lr_pipeline,
    param_grid=lr_params,
    cv=cv,
    scoring="f1_weighted",
    n_jobs=-1
)

lr_grid.fit(X_train, y_train)

print("Best parameters:")
print(lr_grid.best_params_)

print("\nBest CV F1-score:")
print(round(lr_grid.best_score_, 4))
````

````python
# ============================================
# STEP 11.3: KNN TUNING
# ============================================

knn_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", KNeighborsClassifier())
])

knn_params = {
    "model__n_neighbors": [3, 5, 7, 9, 11],
    "model__weights": ["uniform", "distance"],
    "model__metric": ["euclidean", "manhattan"]
}

knn_grid = GridSearchCV(
    estimator=knn_pipeline,
    param_grid=knn_params,
    cv=cv,
    scoring="f1_weighted",
    n_jobs=-1
)

knn_grid.fit(X_train, y_train)

print("Best parameters:")
print(knn_grid.best_params_)

print("\nBest CV F1-score:")
print(round(knn_grid.best_score_, 4))
````

````python
# ============================================
# STEP 11.4: RANDOM FOREST TUNING
# ============================================

rf_pipeline = Pipeline([
    ("model", RandomForestClassifier(random_state=42))
])

rf_params = {
    "model__n_estimators": [50, 100, 200],
    "model__max_depth": [None, 3, 5, 10],
    "model__min_samples_split": [2, 5],
    "model__min_samples_leaf": [1, 2]
}

rf_grid = GridSearchCV(
    estimator=rf_pipeline,
    param_grid=rf_params,
    cv=cv,
    scoring="f1_weighted",
    n_jobs=-1
)

rf_grid.fit(X_train, y_train)

print("Best parameters:")
print(rf_grid.best_params_)

print("\nBest CV F1-score:")
print(round(rf_grid.best_score_, 4))
````

````python
# ============================================
# STEP 11.5: BEST TUNED MODELS
# ============================================

tuned_models = {
    "Tuned Logistic Regression": lr_grid.best_estimator_,
    "Tuned KNN": knn_grid.best_estimator_,
    "Tuned Random Forest": rf_grid.best_estimator_
}

print("Tuned models:")
for name in tuned_models:
    print("-", name)
````

````python
# ============================================
# STEP 11.6: FINAL TEST EVALUATION
# ============================================

tuned_results = []

for name, model in tuned_models.items():

    y_pred = model.predict(X_test)

    tuned_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test,
            y_pred,
            average="weighted"
        ),
        "Recall": recall_score(
            y_test,
            y_pred,
            average="weighted"
        ),
        "F1-Score": f1_score(
            y_test,
            y_pred,
            average="weighted"
        )
    })

tuned_results_df = pd.DataFrame(tuned_results)

tuned_results_df[
    ["Accuracy", "Precision", "Recall", "F1-Score"]
] = tuned_results_df[
    ["Accuracy", "Precision", "Recall", "F1-Score"]
].round(4)

tuned_results_df
````

````python
# ============================================
# STEP 11.7: BASELINE VS TUNED
# ============================================

baseline_results = results_df.copy()

baseline_results["Type"] = "Baseline"

tuned_comparison = tuned_results_df.copy()
tuned_comparison["Type"] = "Tuned"

combined_results = pd.concat(
    [baseline_results, tuned_comparison],
    ignore_index=True
)

combined_results
````

## 🔧 Hyperparameter Optimization

Hyperparameter optimization was performed using GridSearchCV with
Stratified 5-Fold Cross-Validation.

Logistic Regression, K-Nearest Neighbors, and Random Forest were

The test set was kept separate during hyperparameter optimization

Baseline and tuned models were compared to determine whether

````python
# ============================================
# STEP 12.1: COMBINE MODEL RESULTS
# ============================================

all_test_results = pd.concat(
    [
        results_df.assign(Type="Baseline"),
        tuned_results_df.assign(Type="Tuned")
    ],
    ignore_index=True
)

all_test_results
````

````python
# ============================================
# STEP 12.2: FINAL MODEL RANKING
# ============================================

final_ranking = all_test_results.sort_values(
    by=["F1-Score", "Accuracy", "Precision", "Recall"],
    ascending=False
).reset_index(drop=True)

final_ranking
````

````python
# ============================================
# STEP 12.3: SELECT BEST MODEL
# ============================================

best_model_name = final_ranking.loc[0, "Model"]

print("🏆 FINAL SELECTED MODEL")
print("-----------------------")
print("Model:", best_model_name)
print("Accuracy:", final_ranking.loc[0, "Accuracy"])
print("Precision:", final_ranking.loc[0, "Precision"])
print("Recall:", final_ranking.loc[0, "Recall"])
print("F1-Score:", final_ranking.loc[0, "F1-Score"])
````

````python
# ============================================
# STEP 12.4: GET FINAL MODEL OBJECT
# ============================================

all_models = {}

# Baseline models
all_models.update(trained_models)

# Tuned models
all_models.update(tuned_models)

final_model = all_models[best_model_name]

print("Final model object:")
print(final_model)
````

````python
# ============================================
# STEP 12.5: FINAL MODEL PREDICTION
# ============================================

final_predictions = final_model.predict(X_test)

final_accuracy = accuracy_score(
    y_test,
    final_predictions
)

print("Final Test Accuracy:", round(final_accuracy, 4))
````

````python
# ============================================
# STEP 12.6: FINAL CONFUSION MATRIX
# ============================================

final_cm = confusion_matrix(
    y_test,
    final_predictions,
    labels=iris.target_names
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    final_cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)

plt.title(f"Confusion Matrix - {best_model_name}")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.tight_layout()
plt.show()
````

````python
# ============================================
# STEP 12.7: FINAL CLASSIFICATION REPORT
# ============================================

print(
    classification_report(
        y_test,
        final_predictions
    )
)
````

````python
# ============================================
# STEP 14.1: ERROR ANALYSIS DATAFRAME
# ============================================

error_analysis = X_test.copy()

error_analysis["Actual"] = y_test.values
error_analysis["Predicted"] = final_predictions

error_analysis["Correct"] = (
    error_analysis["Actual"] ==
    error_analysis["Predicted"]
)

error_analysis.head()
````

````python
# ============================================
# STEP 14.2: MISCLASSIFIED SAMPLES
# ============================================

misclassified = error_analysis[
    error_analysis["Correct"] == False
]

print("Total test samples:", len(error_analysis))
print("Misclassified samples:", len(misclassified))
print(
    "Error rate:",
    round(len(misclassified) / len(error_analysis), 4)
)
````

````python
# ============================================
# STEP 14.3: VIEW ERRORS
# ============================================

misclassified
````

````python
# ============================================
# STEP 14.4: ERROR PATTERN
# ============================================

error_pattern = pd.crosstab(
    error_analysis["Actual"],
    error_analysis["Predicted"],
    rownames=["Actual"],
    colnames=["Predicted"]
)

error_pattern
````

````python
# ============================================
# STEP 14.5: CLASS-WISE ERROR RATE
# ============================================

class_error_rate = (
    error_analysis
    .groupby("Actual")["Correct"]
    .apply(lambda x: 1 - x.mean())
    .reset_index(name="Error Rate")
)

class_error_rate["Error Rate"] = (
    class_error_rate["Error Rate"].round(4)
)

class_error_rate
````

````python
# ============================================
# STEP 14.6: CLASS-WISE ERROR VISUALIZATION
# ============================================

plt.figure(figsize=(9, 5))

sns.barplot(
    data=class_error_rate,
    x="Actual",
    y="Error Rate"
)

plt.title("Class-wise Error Rate")
plt.xlabel("Actual Species")
plt.ylabel("Error Rate")

plt.ylim(0, 1)

plt.tight_layout()
plt.show()
````

````python
# ============================================
# STEP 14.8: VISUALIZE MISCLASSIFICATIONS
# ============================================

plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=error_analysis,
    x="petal length (cm)",
    y="petal width (cm)",
    hue="Actual",
    style="Correct",
    s=100
)

plt.title(
    "Actual Species and Classification Errors"
)

plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")

plt.tight_layout()
plt.show()
````
